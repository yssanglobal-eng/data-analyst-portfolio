```python
# GRÁFICA 1 — Ingresos totales por tipo de tratamiento
ingresos = df_tratamientos.groupby('tipo')['costo'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=ingresos.values, y=ingresos.index, palette='Blues_r')
plt.title('Ingresos totales por tipo de tratamiento', fontsize=14, fontweight='bold')
plt.xlabel('Ingreso total (MXN)')
plt.ylabel('Tipo de tratamiento')
plt.tight_layout()
plt.show()
```
